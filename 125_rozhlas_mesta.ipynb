```python
import polars as pl
```

```python
pl.Config(tbl_rows=200)
pl.Config.set_fmt_str_lengths(200)
pl.Config.set_tbl_width_chars(200)
```

```python
df = pl.read_parquet(
    "data/cro.parquet"
).with_columns(
    pl.col("track").str.to_lowercase()
).with_columns(
    pl.col("interpret").str.to_lowercase()
).with_columns(
    (pl.col('interpret') + ": " + pl.col("track")).alias("banger")
).with_columns(
    pl.col('since').dt.week().alias('tyden')
)
```

```python
df.select(pl.col("since")).min()
```

```python
df.select(pl.col("since")).max()
```
